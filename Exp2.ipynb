```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
print("libraries imported successfully")
data={
    'student': ['s1','s2','s3','s4','s5','s6','s7','s8','s9','s10','s11'],
    'age':[20,21,20,22,21,23,20,22,21,23,25],
    'attendance':[85,90,78,92,88,70,95,82,76,89,90],
    'marks': [78,85,72,91,80,65,64,88,300,76,70],
    'study_hours':[3,4,5,2,3,2,6,3,4,6,7],
    'dept':['cse','eee','cse','ece','cse','cse','eee','cse','cse','cse','eee'],
}
```

```python
df=pd.DataFrame(data)
print(df)
```

```python
print("First Five Records")
print(df.head())

print("Dataset Shape:")
print(df.shape)

print("Data Types:")
print(df.dtypes)

print("Statistical Summary:")
print(df.describe())

print("Last Five Records")
print(df.tail())
```

```python
department_marks = df.groupby("dept")["marks"].mean()
plt.figure(figsize=(8, 5))
plt.bar(department_marks.index, department_marks.values)
plt.title("Average Marks by Department")
plt.xlabel("Department")
plt.ylabel("Average Marks")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.plot(df["student"], df["marks"], marker='o')
plt.title("Student Marks")
plt.xlabel("Students")
plt.ylabel("Marks")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.hist(df["marks"], bins=5, edgecolor='black')
plt.title("Distribution of Marks")
plt.xlabel("Marks")
plt.ylabel("Frequency")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(df["study_hours"], df["marks"])
plt.title("Study Hours vs Marks")
plt.xlabel("Study Hours")
plt.ylabel("Marks")
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.boxplot(df["marks"], vert=True)
plt.title("Box Plot")
plt.ylabel("Marks")
plt.show()
```

```python
department_count = df["dept"].value_counts()
plt.figure(figsize=(7, 7))
plt.pie(department_count.values, labels=department_count.index, autopct='%1.1f%%', startangle=90)
plt.title("Students by Department")
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

numeric_data = df.select_dtypes(include=np.number)

correlation_matrix = numeric_data.corr()

print(correlation_matrix)

plt.figure(figsize=(8, 6))

sns.heatmap(
    correlation_matrix,
    annot=True,
    fmt='.2f',
    cmap="coolwarm"
)

plt.title("Correlation Heatmap")

plt.show()
```
