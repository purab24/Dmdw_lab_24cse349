```python
lst = [34, 43, 81, 106, 106, 115]

sum = 0

for i in range(len(lst)):
    sum += int(lst[i])

mean = sum / len(lst)

print("Mean is:", mean)
```

```python
x = [10, 20, 30, 40, 50, 60]
y = [2, 3, 4, 10, 5, 8]

tot_sum = 0

for i in range(len(x)):
    tot_sum += x[i] * y[i]

freq_tot = 0
for val in y:
    freq_tot += val

mean = tot_sum / freq_tot

print("Mean is:", mean)
```

```python
df["Age"] = df["Age"].fillna(df["Age"].mean())
df["Salary"] = df["Salary"].fillna(df["Salary"].median())
print(df.isnull().sum())
```

```python
df = df.drop_duplicates()

print(df)
```

```python
q1 = df["Salary"].quantile(0.25)
q3 = df["Salary"].quantile(0.75)
print("Q1:", q1)
print("Q3:", q3)
```

```python
q1 = df["Salary"].quantile(0.25)
q3 = df["Salary"].quantile(0.75)
IQR = q3 - q1
print("IQR:", IQR)
```

```python
q1 = df["Salary"].quantile(0.25)
q3 = df["Salary"].quantile(0.75)
IQR = q3 - q1
lower = q1 - 1.5 * IQR
upper = q3 + 1.5 * IQR
print("Lower Bound:", lower)
print("Upper Bound:", upper)
```

```python
q1_salary = df["Salary"].quantile(0.25)
q3_salary = df["Salary"].quantile(0.75)
iqr_salary = q3_salary - q1_salary
lower_salary = q1_salary - 1.5 * iqr_salary
upper_salary = q3_salary + 1.5 * iqr_salary
outliers = df[(df["Salary"] < lower_salary) | (df["Salary"] > upper_salary)]
print("Outliers:", outliers)
```
